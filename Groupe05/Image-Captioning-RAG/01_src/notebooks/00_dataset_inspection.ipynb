## 4. Dataset Directory Inspection

This section examines the directory organisation of the processed dataset.

The objective is to verify the presence of the training, validation, and test splits, including their image and annotation directories.

```python

# Define expected directories and files
dataset_structure = {
    "training": {
        "images": TRAINING_ROOT / "images",
        "annotations": TRAINING_ROOT / "annotations",
        "captions": TRAINING_ROOT / "annotations" / "captions.json",
        "instances": TRAINING_ROOT / "annotations" / "instances.json",
    },
    "validation": {
        "images": VALIDATION_ROOT / "images",
        "annotations": VALIDATION_ROOT / "annotations",
        "captions": VALIDATION_ROOT / "annotations" / "captions.json",
        "instances": VALIDATION_ROOT / "annotations" / "instances.json",
    },
    "test": {
        "images": TEST_ROOT / "images",
        "annotations": TEST_ROOT / "annotations",
        "captions": TEST_ROOT / "annotations" / "captions.json",
        "instances": TEST_ROOT / "annotations" / "instances.json",
    },
}

# Display directory and file availability
for split_name, paths in dataset_structure.items():
    print(f"\n{'=' * 50}")
    print(f"{split_name.upper()} SPLIT")
    print("=" * 50)

    for item_name, path in paths.items():
        status = "FOUND" if path.exists() else "MISSING"
        print(f"{item_name:<15}: {status}")
```

## 5. COCO Annotation Structure Inspection

This section examines the structure of the COCO annotation files for the training, validation, and test splits.

The objective is to:
- Identify the main keys available in each JSON file.
- Inspect representative image records.
- Inspect representative caption and object annotation records.
- Examine the category information available in the instance annotations.

This inspection helps identify which metadata can be used in the subsequent statistical exploration.

```python
# Load annotation JSON files for each split

annotation_files = {
    "training_captions": TRAINING_ROOT / "annotations" / "captions.json",
    "training_instances": TRAINING_ROOT / "annotations" / "instances.json",

    "validation_captions": VALIDATION_ROOT / "annotations" / "captions.json",
    "validation_instances": VALIDATION_ROOT / "annotations" / "instances.json",

    "test_captions": TEST_ROOT / "annotations" / "captions.json",
    "test_instances": TEST_ROOT / "annotations" / "instances.json",
}

annotations_data = {}

for name, path in annotation_files.items():
    with open(path, "r", encoding="utf-8") as file:
        annotations_data[name] = json.load(file)

    print(f"{name}: loaded successfully")
```

```python
# Display the top-level keys in each annotation file

for name, data in annotations_data.items():
    print(f"\n{name.upper()}")
    print("-" * 50)

    for key in data.keys():
        value = data[key]

        if isinstance(value, list):
            print(f"{key:<20}: list ({len(value)} elements)")
        elif isinstance(value, dict):
            print(f"{key:<20}: dictionary ({len(value)} keys)")
        else:
            print(f"{key:<20}: {type(value).__name__}")
```

```python
# Inspect representative records from the training split

training_captions = annotations_data["training_captions"]
training_instances = annotations_data["training_instances"]

print("IMAGE RECORD")
print(training_captions["images"][0])

print("\nCAPTION RECORD")
print(training_captions["annotations"][0])

print("\nOBJECT ANNOTATION RECORD")
print(training_instances["annotations"][0])

print("\nCATEGORY RECORD")
print(training_instances["categories"][0])
```
